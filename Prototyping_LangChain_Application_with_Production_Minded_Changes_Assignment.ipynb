#### Production Caching Architecture

Our LLMOps library implements sophisticated caching at multiple levels:

**Embedding Caching:**
The process of embedding is typically very time consuming and expensive:

1. Send text to OpenAI API endpoint
2. Wait for processing  
3. Receive response
4. Pay for API call

This occurs *every single time* a document gets converted into a vector representation.

**Our Caching Solution:**
1. Check local cache for previously computed embeddings
2. If found: Return cached vector (instant, free)
3. If not found: Call OpenAI API, store result in cache
4. Return vector representation

**LLM Response Caching:**
Similarly, we cache LLM responses to avoid redundant API calls for identical prompts.

**Benefits:**
- ⚡ Faster response times (cache hits are instant)
- 💰 Reduced API costs (no duplicate calls)  
- 🔄 Consistent results for identical inputs
- 📈 Better scalability

Our ProductionRAGChain automatically handles all this caching behind the scenes!

##### ✅ Answer: Production Caching Analysis

**Limitations of this caching approach:**

1. **Memory vs Disk caching trade-offs:**
   - Current in-memory cache is fast but volatile - data lost on restart
   - No persistence across application restarts
   - Memory usage grows with cache size, potentially causing OOM errors

2. **Cache invalidation strategies:**
   - No automatic cache expiration or TTL (Time To Live)
   - Stale data remains cached indefinitely
   - No versioning for document updates

3. **Concurrent access patterns:**
   - In-memory cache may not handle high concurrent access well
   - Race conditions possible with multiple processes
   - No distributed caching for multi-instance deployments

4. **Cache size management:**
   - No LRU (Least Recently Used) eviction policy
   - Cache grows unbounded, consuming memory
   - No compression for large embeddings

5. **Cold start scenarios:**
   - First-time users experience full latency
   - No pre-warming of common queries
   - Cache miss on every new deployment

**Most useful for production when:**
- Application has stable, predictable query patterns
- Memory resources are sufficient for expected cache size
- Single-instance deployment or simple scaling
- Real-time performance is critical

**Least useful for production when:**
- High availability requirements (multi-instance)
- Memory constraints exist
- Query patterns are highly dynamic
- Long-term data persistence needed
- Need for cache sharing across instances

#### Adding Cache Hit Rate Analysis Test below

````python
##### 🧮 Cache Hit Rate Analysis

# Assumes `rag_chain`, `test_texts`, and `test_questions` are already defined above
# Strategy: one warm-up miss per unique input, then R repeated calls counted as hits.

R = 2  # number of repeated (cached) calls after the warm-up

# Embeddings: 1 unique input (test_texts)
embedding_total = 0
embedding_hits = 0

# Warm-up (miss)
_ = rag_chain.cached_embeddings.get_embeddings().embed_documents(test_texts)
embedding_total += 1

# Repeats (hits)
for _ in range(R):
    _ = rag_chain.cached_embeddings.get_embeddings().embed_documents(test_texts)
    embedding_total += 1
    embedding_hits += 1

# LLM: warm-up once per unique question, then repeat R times (hits)
llm_total = 0
llm_hits = 0

# Warm-up (misses): one per unique question
for q in test_questions:
    _ = rag_chain.invoke(q)
    llm_total += 1

# Repeats (hits): R per question
for _ in range(R):
    for q in test_questions:
        _ = rag_chain.invoke(q)
        llm_total += 1
        llm_hits += 1

embedding_hit_rate = (embedding_hits / embedding_total) * 100 if embedding_total else 0.0
llm_hit_rate = (llm_hits / llm_total) * 100 if llm_total else 0.0
overall_hit_rate = ((embedding_hits + llm_hits) / (embedding_total + llm_total)) * 100 if (embedding_total + llm_total) else 0.0

print("\n📊 Cache Hit Rate Analysis:")
print("========================================")
print(f"Embedding Cache Hit Rate: {embedding_hit_rate:.1f}%  "
      f"({embedding_hits}/{embedding_total})")
print(f"LLM Cache Hit Rate:       {llm_hit_rate:.1f}%  "
      f"({llm_hits}/{llm_total})")
print(f"Overall Cache Hit Rate:   {overall_hit_rate:.1f}%  "
      f"({embedding_hits + llm_hits}/{embedding_total + llm_total})")
````

##### ✅ Answer:

````python
### YOUR EXPERIMENTATION CODE HERE ###

import os, time
from pathlib import Path
from langchain_core.messages import HumanMessage, AIMessage

def create_helpfulness_agent(rag_chain, model_name="gpt-4.1-mini", temperature=0.1):
    from typing_extensions import TypedDict, Annotated
    from langgraph.graph.message import add_messages
    from langgraph.graph import StateGraph, END
    from langgraph.prebuilt import ToolNode
    from langchain_core.prompts import PromptTemplate
    from langchain_core.output_parsers import StrOutputParser
    from langchain_core.messages import ToolMessage
    from langgraph_agent_lib.agents import get_default_tools
    from langgraph_agent_lib import get_openai_model

    class AgentState(TypedDict):
        messages: Annotated[list, add_messages]

    tools = get_default_tools(rag_chain)
    model = get_openai_model(model_name=model_name, temperature=temperature)
    model_with_tools = model.bind_tools(tools)

    def call_model(state: AgentState):
        messages = state["messages"]
        response = model_with_tools.invoke(messages)
        return {"messages": [response]}

    def route_to_action_or_helpfulness(state: AgentState):
        last_message = state["messages"][-1]
        if getattr(last_message, "tool_calls", None):
            return "action"
        return "helpfulness"

    def _tools_used_in_state(state: AgentState) -> set[str]:
        names: set[str] = set()
        for m in state["messages"]:
            if isinstance(m, ToolMessage) and getattr(m, "name", None):
                names.add(m.name)
            if hasattr(m, "tool_calls") and m.tool_calls:
                for tc in m.tool_calls:
                    n = tc.get("name") if isinstance(tc, dict) else getattr(tc, "name", None)
                    if n:
                        names.add(n)
        return names

    def helpfulness_node(state: AgentState):
        if len(state["messages"]) > 20:
            return {"messages": [AIMessage(content="HELPFULNESS:END")]}

        initial_msg = state["messages"][0]
        last_msg = state["messages"][-1]

        initial_text = getattr(initial_msg, "content", str(initial_msg)) or ""
        response_content = getattr(last_msg, "content", str(last_msg)) or ""

        is_complex = any(k in initial_text.lower() for k in ["relate", "compare", "synthesize", "multi", "trend"])
        used_tools = _tools_used_in_state(state)

        if is_complex and (len(used_tools) < 2 or len(response_content) < 800):
            feedback = (
                "Use multiple tools (retrieve_information, tavily_search_results_json, arxiv) to gather evidence "
                "and produce a longer, well-structured synthesis that explicitly relates the concepts."
            )
            return {"messages": [AIMessage(content="HELPFULNESS:N"), HumanMessage(content=feedback)]}

        prompt = PromptTemplate.from_template(
            "Given an initial query and a final response, decide if the response is extremely helpful. "
            "Reply with 'Y' for helpful and 'N' for unhelpful.\n\nInitial Query:\n{initial_query}\n\nFinal Response:\n{final_response}"
        )
        judge = get_openai_model(model_name="gpt-4.1-mini", temperature=0)
        chain = prompt | judge | StrOutputParser()
        decision_text = chain.invoke({"initial_query": initial_text, "final_response": response_content})
        decision = "Y" if "Y" in (decision_text or "") else "N"
        return {"messages": [AIMessage(content=f"HELPFULNESS:{decision}")]}  # marker

    def helpfulness_decision(state: AgentState):
        last_marker = None
        for m in reversed(state["messages"]):
            c = getattr(m, "content", "")
            if isinstance(c, str) and c.startswith("HELPFULNESS:"):
                last_marker = c
                break
        if last_marker == "HELPFULNESS:END":
            return END
        if last_marker == "HELPFULNESS:Y":
            return "end"
        return "continue"

    graph = StateGraph(AgentState)
    tool_node = ToolNode(tools)
    graph.add_node("agent", call_model)
    graph.add_node("action", tool_node)
    graph.add_node("helpfulness", helpfulness_node)
    graph.set_entry_point("agent")
    graph.add_conditional_edges("agent", route_to_action_or_helpfulness,
                                {"action": "action", "helpfulness": "helpfulness"})
    graph.add_conditional_edges("helpfulness", helpfulness_decision,
                                {"continue": "agent", "end": END, END: END})
    graph.add_edge("action", "agent")
    return graph.compile()

# Test queries
queries_to_test = [
    "What is the main purpose of the Direct Loan Program?",
    "What are the latest developments in AI safety?",
    "Find recent papers about transformer architectures",
    "How do the concepts in this document relate to current AI research trends?"
]

# Create helpfulness agent
helpfulness_agent = create_helpfulness_agent(rag_chain)

# Helpers to show real content in prod tests
def extract_final_content(run):
    msgs = run.get("messages", []) if isinstance(run, dict) else []
    for m in reversed(msgs):
        c = getattr(m, "content", "")
        if isinstance(c, str) and not c.startswith("HELPFULNESS:"):
            return c
    return getattr(msgs[-1], "content", "") if msgs else ""

def show_first_200(resp):
    try:
        txt = extract_final_content(resp) or ""
        print("   Response: " + txt[:200].replace("\n", " "))
    except Exception:
        pass

# Run side-by-side tests
comprehensive_results = []

print("🚀 Enhanced Agent Testing with Tool Analysis and Performance Metrics")
print("=" * 80)

for i, query in enumerate(queries_to_test, 1):
    print(f"\n🔍 Test {i}: {query}")
    print("-" * 60)

    # Simple Agent
    print("🤖 Simple Agent Analysis:")
    print("  Testing...")
    t0 = time.time()
    try:
        s_msgs = [HumanMessage(content=query)]
        s_resp = simple_agent.invoke({"messages": s_msgs})
        s_time = time.time() - t0

        s_tools = []
        for m in s_resp["messages"]:
            if hasattr(m, "tool_calls") and m.tool_calls:
                for tc in m.tool_calls:
                    s_tools.append(tc.get("name") if isinstance(tc, dict) else getattr(tc, "name", None))

        s_final = s_resp["messages"][-1].content
        print(f"  ✅ Response Time: {s_time:.3f}s")
        print(f"  📊 Messages Used: {len(s_resp['messages'])}")
        print(f"  🛠️  Tools Used: {s_tools if s_tools else 'None'}")
        print(f"  Response Length: {len(s_final)} characters")
        print("  Final Response:")
        print("    " + s_final.replace("\n", "\n    "))
        s_ok = True
    except Exception as e:
        print(f"  ❌ Error: {e}")
        s_time, s_tools, s_final, s_ok = float("inf"), [], f"Error: {e}", False

    # Helpfulness Agent
    print("\n🧠 Helpfulness Agent Analysis:")
    print("  Testing...")
    t0 = time.time()
    try:
        h_msgs = [HumanMessage(content=query)]
        h_resp = helpfulness_agent.invoke({"messages": h_msgs})
        h_time = time.time() - t0

        h_tools = []
        for m in h_resp["messages"]:
            if hasattr(m, "tool_calls") and m.tool_calls:
                for tc in m.tool_calls:
                    h_tools.append(tc.get("name") if isinstance(tc, dict) else getattr(tc, "name", None))

        h_final = ""
        h_evals = 0
        for m in h_resp["messages"]:
            c = getattr(m, "content", "")
            if isinstance(c, str) and c.startswith("HELPFULNESS:"):
                h_evals += 1
            else:
                h_final = c

        print(f"  ✅ Response Time: {h_time:.3f}s")
        print(f"  📊 Messages Used: {len(h_resp['messages'])}")
        print(f"  🛠️  Tools Used: {h_tools if h_tools else 'None'}")
        print(f"  Response Length: {len(h_final)} characters")
        print(f"  🔄 Helpfulness Evaluations: {h_evals}")
        print("  Final Response:")
        print("    " + str(h_final).replace("\n", "\n    "))
        h_ok = True
    except Exception as e:
        print(f"  ❌ Error: {e}")
        h_time, h_tools, h_final, h_evals, h_ok = float("inf"), [], f"Error: {e}", 0, False

    comprehensive_results.append({
        "query": query,
        "query_type": ["RAG", "Web", "Academic", "Multi-tool"][i-1],
        "simple_agent": {
            "success": s_ok,
            "response_time": s_time,
            "message_count": len(s_resp["messages"]) if s_ok else 0,
            "tools_used": s_tools,
            "response_length": len(s_final) if isinstance(s_final, str) else 0,
            "final_response": s_final
        },
        "helpfulness_agent": {
            "success": h_ok,
            "response_time": h_time,
            "message_count": len(h_resp["messages"]) if h_ok else 0,
            "tools_used": h_tools,
            "response_length": len(h_final) if isinstance(h_final, str) else 0,
            "evaluation_count": h_evals,
            "final_response": h_final
        }
    })

    print("\n" + "="*80)

# Analysis
print("\n📊 COMPREHENSIVE ANALYSIS RESULTS:")
print("=" * 80)

print("\n🛠️  TOOL SELECTION PATTERN ANALYSIS:")
print("-" * 50)
summary = {}
for r in comprehensive_results:
    qt = r["query_type"]
    summary.setdefault(qt, {"simple": [], "helpful": []})
    summary[qt]["simple"].extend(r["simple_agent"]["tools_used"])
    summary[qt]["helpful"].extend(r["helpfulness_agent"]["tools_used"])
for qt, tools in summary.items():
    print(f"\n{qt} Queries:")
    print(f"  Simple Agent: {list(set(tools['simple']))}")
    print(f"  Helpful Agent: {list(set(tools['helpful']))}")

print("\n⚡ PERFORMANCE ANALYSIS:")
print("-" * 50)
for r in comprehensive_results:
    if r["simple_agent"]["success"] and r["helpfulness_agent"]["success"]:
        s, h = r["simple_agent"]["response_time"], r["helpfulness_agent"]["response_time"]
        winner = "Simple Agent" if s < h else "Helpfulness Agent"
        speed = (h/s) if s < h else (s/h)
        print(f"\n🔍 {r['query_type']}: {r['query'][:50]}...")
        print(f"  🏆 Winner: {winner} ({speed:.1f}x faster)")
        print(f"  ⏱️  Simple: {s:.3f}s | Helpful: {h:.3f}s")
        print(f"  📊 Simple: {r['simple_agent']['message_count']} msgs | Helpful: {r['helpfulness_agent']['message_count']} msgs")

print("\n🧠 HELPFULNESS EVALUATION ANALYSIS:")
print("-" * 50)
total_eval = sum(r["helpfulness_agent"]["evaluation_count"] for r in comprehensive_results)
avg_eval = total_eval / len(comprehensive_results) if comprehensive_results else 0
print(f"📊 Total Helpfulness Evaluations: {total_eval}")
print(f"📊 Average Evaluations per Query: {avg_eval:.1f}")
print(f"📊 Evaluation Overhead: {avg_eval - 1:.1f} extra messages per query")

# Enhanced cache performance
print("\n📊 Enhanced Cache Performance Analysis:")
print("=" * 50)
test_query = "What are the loan limits for Direct Loans?"

print(f"1️⃣ Testing repeated identical queries: {test_query}")
print("\n Simple Agent Cache Performance:")
t0 = time.time(); _ = simple_agent.invoke({"messages": [HumanMessage(content=test_query)]}); a1 = time.time()-t0
t0 = time.time(); _ = simple_agent.invoke({"messages": [HumanMessage(content=test_query)]}); a2 = time.time()-t0
print(f"First call (cache miss): {a1:.3f}s")
print(f"Cached call (cache hit): {a2:.3f}s")
simple_speedup = a1/a2 if a2>0 else float("inf")
print(f"Simple Agent Cache speedup: {simple_speedup:.1f}x faster!")

print(f"\n2️⃣ Testing variations of similar queries:")
variants = [
    "What are the loan limits for Direct Loans?",
    "What are the maximum loan amounts for Direct Loans?",
    "How much can I borrow with Direct Loans?",
    "What are the Direct Loan borrowing limits?"
]
print("\n🔄 Similar but different queries:")
for i, q in enumerate(variants):
    t0 = time.time(); _ = simple_agent.invoke({"messages": [HumanMessage(content=q)]}); dt = time.time()-t0
    print(f"  Query {i+1}: {q[:50]}... → {dt:.3f}s")

# Correct cache paths (unchanged)
print(f"\n3️⃣ Monitoring cache directory growth:")
paths_to_check = [
    ("📁 Embeddings cache", "/home/saimo/repos/Assignment/16_Production_RAG_and_Guardrails/cache/embeddings"),
    ("📄 LLM sqlite cache", "/home/saimo/repos/Assignment/16_Production_RAG_and_Guardrails/cache/llm_cache.db"),
]
for label, p in paths_to_check:
    if os.path.isdir(p):
        total = sum(f.stat().st_size for f in Path(p).rglob("*") if f.is_file())
        print(f"  {label}: {total/(1024*1024):.2f} MB ({p})")
    elif os.path.isfile(p):
        print(f"  {label}: {os.path.getsize(p)/(1024*1024):.2f} MB ({p})")
    else:
        print(f"  {label}: Not found ({p})")
        if p.endswith("llm_cache.db"):
            print("     Hint: enable disk LLM cache with setup_llm_cache(cache_type='sqlite', cache_path='./cache/llm_cache.db')")

print(f"\n🧠 Helpfulness Agent Cache Performance:")
t0 = time.time(); _ = helpfulness_agent.invoke({"messages": [HumanMessage(content=test_query)]}); b1 = time.time()-t0
t0 = time.time(); _ = helpfulness_agent.invoke({"messages": [HumanMessage(content=test_query)]}); b2 = time.time()-t0
print(f"First call (cache miss): {b1:.3f}s")
print(f"Cached call (cache hit): {b2:.3f}s")
helpful_speedup = b1/b2 if b2>0 else float("inf")
print(f"Helpfulness Agent Cache speedup: {helpful_speedup:.1f}x faster!")
print(f"\n🏆 Cache Performance Winner: {'Simple Agent' if simple_speedup>helpful_speedup else 'Helpfulness Agent'}")

# Production readiness tests (show real content)
print("\n🏭 Enhanced Production Readiness Testing:")
print("=" * 50)

print("\n🤖 Simple Agent Production Tests:")
print("1️⃣ Invalid query (empty):")
try:
    r = simple_agent.invoke({"messages": [HumanMessage(content="")]})
    print("✅ Handled gracefully"); show_first_200(r)
except Exception as e:
    print(f"❌ {e}")

print("2️⃣ RAG tool failure probe (non-existent doc):")
try:
    r = simple_agent.invoke({"messages": [HumanMessage(content="Find information about non-existent document")]})
    print("✅ Handled gracefully"); show_first_200(r)
except Exception as e:
    print(f"❌ {e}")

print("3️⃣ Tool availability (may fail if API keys missing):")
try:
    r = simple_agent.invoke({"messages": [HumanMessage(content="Search for something that might fail")]})
    print("✅ Tools handled gracefully"); show_first_200(r)
except Exception as e:
    print(f"❌ {e}")

print("\n🧠 Helpfulness Agent Production Tests:")
print("1️⃣ Invalid query (empty):")
try:
    r = helpfulness_agent.invoke({"messages": [HumanMessage(content="")]})
    print("✅ Handled gracefully"); show_first_200(r)
except Exception as e:
    print(f"❌ {e}")

print("2️⃣ RAG tool failure probe (non-existent doc):")
try:
    r = helpfulness_agent.invoke({"messages": [HumanMessage(content="Find information about non-existent document")]})
    print("✅ Handled gracefully"); show_first_200(r)
except Exception as e:
    print(f"❌ {e}")

print("3️⃣ Tool availability (may fail if API keys missing):")
try:
    r = helpfulness_agent.invoke({"messages": [HumanMessage(content="Search for something that might fail")]})
    print("✅ Tools handled gracefully"); show_first_200(r)
except Exception as e:
    print(f"❌ {e}")

print("\n📝 Response quality spot-check:")
q = "What are the key benefits of federal student loans?"
r = helpfulness_agent.invoke({"messages": [HumanMessage(content=q)]})
final_txt = extract_final_content(r)
print("✅ Adequate" if len(final_txt) > 100 else "❌ Too short")

print("\n🔍 LangSmith Tracing and Monitoring:")
print("=" * 50)
try:
    import langsmith
    if os.getenv("LANGCHAIN_TRACING_V2") == "true" and os.getenv("LANGCHAIN_API_KEY"):
        project = os.getenv("LANGCHAIN_PROJECT", "Production RAG and Guardrails")
        print(f"✅ Tracing configured for project: {project}")
        try:
            with langsmith.trace(project_name=project, name="Simple Agent Test") as tr:
                _ = simple_agent.invoke({"messages": [HumanMessage(content="Test tracing query")]})
                print(f"  Simple trace id: {tr.id}")
        except Exception as e:
            print(f"  Simple trace error: {e}")
        try:
            with langsmith.trace(project_name=project, name="Helpfulness Agent Test") as tr:
                _ = helpfulness_agent.invoke({"messages": [HumanMessage(content="Test tracing query")]})
                print(f"  Helpful trace id: {tr.id}")
        except Exception as e:
            print(f"  Helpful trace error: {e}")
        # Exact project URL (restored)
        print("🌐 View traces at: https://smith.langchain.com/o/9cc120d2-ca66-4b3b-9545-ca0497b68deb/projects/p/405ae3fe-a5be-4b44-9cb1-551bd447d496")
    else:
        print("⚠️ Tracing not configured (set LANGCHAIN_TRACING_V2=true and LANGCHAIN_API_KEY).")
except Exception as e:
    print(f"⚠️ LangSmith import skipped: {e}")

print("\n✅ Enhanced Activity #2: Advanced Agent Testing with Tool Analysis Complete!")
````

### 🛡️ What are Guardrails?

Guardrails are specialized validation systems that help "catch" when LLM interactions go outside desired parameters. They operate both **pre-generation** (input validation) and **post-generation** (output validation) to ensure safe, compliant, and on-topic responses.

**Key Categories:**
- **Topic Restriction**: Ensure conversations stay on-topic
- **PII Protection**: Detect and redact sensitive information  
- **Content Moderation**: Filter inappropriate language/content
- **Factuality Checks**: Validate responses against source material
- **Jailbreak Detection**: Prevent adversarial prompt attacks
- **Competitor Monitoring**: Avoid mentioning competitors

### Production Benefits of Guardrails

**🏢 Enterprise Requirements:**
- **Compliance**: Meet regulatory requirements for data protection
- **Brand Safety**: Maintain consistent, appropriate communication tone
- **Risk Mitigation**: Reduce liability from inappropriate AI responses
- **Quality Assurance**: Ensure factual accuracy and relevance

**⚡ Technical Advantages:**
- **Layered Defense**: Multiple validation stages for robust protection
- **Selective Enforcement**: Different guards for different use cases

````python
# Activity #3: Creating and Building a Production-Safe LangGraph Agent with Guardrails
# (Graceful off-topic handling: friendly message + BLOCKED status)

from typing import Dict, Any, List
from langgraph.graph import StateGraph, END
from langchain_core.messages import HumanMessage, AIMessage, BaseMessage, SystemMessage
from langgraph_agent_lib.agents import AgentState, create_langgraph_agent, get_default_tools
from langgraph_agent_lib.rag import ProductionRAGChain
from langchain_community.document_loaders import PyMuPDFLoader

# 0) Build a single RAG over ALL PDFs in ./data by bootstrapping with one file, then adding the rest
pdf_paths = [
    "data/The_Direct_Loan_Program.pdf",
    "data/Applications_and_Verification_Guide.pdf",
    "data/Academic_Calenders_Cost_of_Attendance_and_Packaging.pdf",
    "data/The_Federal_Pell_Grant_Program.pdf",
]

# Bootstrap with the first file
rag_chain = ProductionRAGChain(file_path=pdf_paths[0], cache_dir="./cache")

# Add the remaining PDFs into the same vectorstore (so one retriever covers all docs)
for p in pdf_paths[1:]:
    docs = PyMuPDFLoader(p).load()
    rag_chain.vectorstore.add_documents(docs)  # retriever points to this vectorstore already

# 1) Create the base agent with all tools (RAG/Tavily/Arxiv) bound
base_agent = create_langgraph_agent(
    model_name="gpt-4o-mini",
    temperature=0.2,
    tools=get_default_tools(rag_chain),  # includes retrieve_information when rag_chain is provided
    rag_chain=rag_chain,
)

def _get_text(message: BaseMessage) -> str:
    try:
        return message.content if hasattr(message, "content") else str(message)
    except Exception:
        return str(message)

def _get_latest_user_message(messages: List[BaseMessage]) -> str:
    for m in reversed(messages):
        if isinstance(m, HumanMessage):
            return _get_text(m)
    return ""

def _get_context_from_rag(query: str) -> str:
    if not query:
        return ""
    try:
        docs = rag_chain.retriever.get_relevant_documents(query)
        return "\n\n".join([d.page_content for d in docs]) if docs else ""
    except Exception:
        return ""

GRACEFUL_SCOPE_MSG = (
    "This assistant focuses on student loans. Please ask a loan-related question "
    "(e.g., repayment options, loan limits)."
)

# 2) Input guardrails with graceful handling (topic, jailbreak, PII redaction)
def input_guard_node(state: AgentState) -> Dict[str, Any]:
    last = state["messages"][-1]
    user_text = _get_text(last)
    try:
        # Enforce Topic + Jailbreak
        topic_guard.validate(user_text)
        jailbreak_guard.validate(user_text)

        # PII redaction
        pii_res = pii_guard.validate(user_text)
        redacted = getattr(pii_res, "validated_output", None)
        if redacted and redacted != user_text:
            new_messages = list(state["messages"][:-1]) + [HumanMessage(content=redacted)]
            return {"messages": new_messages}

        return {"messages": state["messages"]}
    except Exception:
        # Graceful scope message; keep BLOCKED prefix so status check still works
        return {"messages": state["messages"] + [AIMessage(content=f"[BLOCKED] {GRACEFUL_SCOPE_MSG}")]}

# 3) RAG context prefetch (inject doc snippets as a system message if available)
def rag_context_node(state: AgentState) -> Dict[str, Any]:
    user_query = _get_latest_user_message(state["messages"])
    context_text = _get_context_from_rag(user_query)
    if context_text:
        ctx_msg = SystemMessage(
            content=("Use the following document context to answer the user. Do not mention this note.\n\n"
                     f"{context_text}")
        )
        return {"messages": state["messages"] + [ctx_msg]}
    return {"messages": state["messages"]}

# 4) Tool-selection policy (nudge agent to choose RAG vs Tavily vs Arxiv)
def tool_policy_node(state: AgentState) -> Dict[str, Any]:
    policy = SystemMessage(content=(
        "Tool-selection policy:\n"
        "- Use `retrieve_information` (RAG) first for questions likely answerable from the PDF/context.\n"
        "- Use `tavily_search_results_json` for current events or web-only info.\n"
        "- Use `arxiv` for academic literature queries (papers, citations, transformer topics).\n"
        "Prefer one tool at a time; cite sources when possible."
    ))
    return {"messages": state["messages"] + [policy]}

# 5) Agent call
def agent_node(state: AgentState) -> Dict[str, Any]:
    res = base_agent.invoke({"messages": state["messages"]})
    return {"messages": res["messages"]}

# 6) Output guardrails (profanity always; factuality only if context exists) with one-shot refinement
def output_guard_node(state: AgentState) -> Dict[str, Any]:
    ai_text = _get_text(state["messages"][-1])
    try:
        profanity_guard.validate(ai_text)
        user_message = _get_latest_user_message(state["messages"])
        context_text = _get_context_from_rag(user_message)
        if context_text.strip():
            factuality_guard.validate(
                ai_text,
                metadata={"user_message": user_message, "context": context_text, "llm_response": ai_text},
            )
        return {"messages": state["messages"]}
    except Exception as e:
        revision_hint = HumanMessage(
            content=f"Revise the previous answer to comply with safety/factuality: {e}. "
                    f"Remove unsafe content and ensure alignment with provided context."
        )
        revised = base_agent.invoke({"messages": state["messages"] + [revision_hint]})
        return {"messages": revised["messages"]}

# 7) Workflow wiring with graceful early-exit
def should_continue_after_input(state: AgentState):
    last_text = _get_text(state["messages"][-1])
    return "rag_context" if not last_text.startswith("[BLOCKED]") else END

graph = StateGraph(AgentState)
graph.add_node("input_guard", input_guard_node)
graph.add_node("rag_context", rag_context_node)
graph.add_node("tool_policy", tool_policy_node)
graph.add_node("agent", agent_node)
graph.add_node("output_guard", output_guard_node)

graph.set_entry_point("input_guard")
graph.add_conditional_edges("input_guard", should_continue_after_input, {"rag_context": "rag_context", END: END})
graph.add_edge("rag_context", "tool_policy")
graph.add_edge("tool_policy", "agent")
graph.add_edge("agent", "output_guard")
graph.add_edge("output_guard", END)

guarded_agent = graph.compile()

# Optional demo:
# from langchain_core.messages import HumanMessage
# result = guarded_agent.invoke({
#     "messages": [HumanMessage(content="Latest developments in AI safety this month?")]
# })
# print(_get_text(result["messages"][-1]))
````
